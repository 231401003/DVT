```python
df = df.drop_duplicates()

print("Duplicates Removed Successfully!")
```

```python
df.to_excel("Churn_Modelling.xlsx", index=False)

print("Excel File Exported Successfully!")
```

```python
excel_df = pd.read_excel("Churn_Modelling.xlsx")

excel_df.head()
```

```python
conn = sqlite3.connect("ChurnDatabase.db")

df.to_sql("CustomerData", conn, if_exists="replace", index=False)

print("Database Created Successfully!")
```

```python
query = "SELECT * FROM CustomerData LIMIT 5"

sql_df = pd.read_sql(query, conn)

sql_df
```

```python
df.to_csv("Churn_Output.csv", index=False)

print("CSV File Exported Successfully!")
```

```python
df.to_json("Churn_Output.json", orient="records", indent=4)

print("JSON File Exported Successfully!")
```

```python
import os

files = [
    "Churn_Modelling_Export.csv",
    "Churn_Modelling_Export.xlsx",
    "Churn_Modelling_Export.json"
]

for file in files:
    if os.path.exists(file):
        print(file, "Created Successfully")
    else:
        print(file, "Not Found")
```
